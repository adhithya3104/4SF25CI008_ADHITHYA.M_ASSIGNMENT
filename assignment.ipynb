```python
#Create a Pandas DataFrame with 3 columns and 4 rows filled with random integers. Compute the row-wise and column-wise sum.

df = pd.DataFrame(np.random.randint(1, 100, size=(4, 3)),
columns=['A', 'B', 'C'])
print(df)
```

```python
#Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. Fill the NaN values with the mean of the respective columns.
df = pd.DataFrame(np.random.randint(1, 100, size=(5, 3)),
columns=['P', 'Q', 'R'])
print(df)
```

```python
df.iloc[1, 0] = np.nan
df.iloc[2, 1] = np.nan
df.iloc[3, 2] = np.nan
df = df.fillna(df.mean())
print("\nDataFrame after filling NaN values:")
print(df)
```

```python
#Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. Introduce some NaN values. Drop the rows with any NaN values

df = pd.DataFrame(np.random.randint(1, 100, size=(6, 4)),
columns=['A', 'B', 'C', 'D'])
print(df)
```

```python
df.iloc[1, 0] = np.nan
df.iloc[2, 1] = np.nan
df.iloc[3, 2] = np.nan
df.iloc[4, 3] = np.nan
df = df.dropna()
print("\nDataFrame after dropping rows with NaN values:")
print(df)
```

```python
#Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category
df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], size=10),
    'Value': np.random.randint(1, 100, size=10)
})
print(df)
grouped = df.groupby('Category')
print("\nSum of values for each category:")
print(grouped['Value'].sum())
print("\nMean of values for each category:")
print(grouped['Value'].mean())
```

```python
#Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. Fill the DataFrame with random data. Group the DataFrame by 'Category' and compute the total sales for each category.

df = pd.DataFrame({
    'Product': np.random.choice(['P1', 'P2', 'P3'], size=10),
    'Category': np.random.choice(['A', 'B', 'C'], size=10),
    'Sales': np.random.randint(1, 100, size=10)
})
grouped = df.groupby('Category')
print("\nTotal sales for each category:")
print(grouped['Sales'].sum())
```

```python
#Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.

df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['virat', 'rohit', 'rahul', 'Dravid']
})
df2 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Age': [25, 30, 35, 40]
})
merged_df = pd.merge(df1, df2, on='ID')
print(merged_df)
```

```python
#Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'. Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.

import pandas as pd
import numpy as np

# Create a MultiIndex
index = pd.MultiIndex.from_product(
    [['A', 'B', 'C'], ['X', 'Y']],
    names=['Category', 'SubCategory']
)
df = pd.DataFrame({
    'Value': np.random.randint(1, 100, size=6)
}, index=index)

print("\nSum by Category:")
print(df.groupby(level='Category').sum())
print("\nSum by SubCategory:")
print(df.groupby(level='SubCategory').sum())
```

```python
df.groupby(level=['Category', 'SubCategory'])['Value'].sum()
```

```python
#Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Apply a function that doubles the values of the DataFrame.

df = pd.DataFrame(np.random.randint(1, 100, size=(5, 3)),
columns=['A', 'B', 'C'])
df = df.apply(lambda x: x * 2)
print("\nDataFrame after doubling:")
print(df)
```

```python
#Create a Pandas DataFrame with 3 columns and 6 rows filled with random integers. Apply a lambda function to create a new column that is the sum of the existing columns.

df = pd.DataFrame(
    np.random.randint(1, 10, size=(6, 3)),
    columns=['A', 'B', 'C']
)

print(df)
```

```python
df['Total'] = df.apply(
    lambda x: x['A'] + x['B'] + x['C'], axis=1
)

print(df)
```
